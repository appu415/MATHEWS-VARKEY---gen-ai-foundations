# Task 1: Vectorized Scaled Dot-Product Attention from Scratch

## Objective

To implement the scaled dot-product attention mechanism from scratch using optimized NumPy operations, with batch-wise multi-head processing, causal sequence masking, and 4D matrix operations.

## Technologies / Tools Used

- Python 3.10+
- NumPy

## Formula

The scaled dot-product attention formula is:

Attention(Q, K, V) = softmax((QKᵀ) / √dₖ)V

Where:

- Q = Query matrix
- K = Key matrix
- V = Value matrix
- Kᵀ = Transpose of the Key matrix
- dₖ = Dimension of the Key vectors
- √dₖ = Scaling factor

## Step 1: Import Required Libraries

Import NumPy for matrix and numerical operations.

A random seed is used to produce consistent results.

```python
# Import NumPy
import numpy as np

# Set random seed for consistent results
np.random.seed(42)
```

## Step 2: Define Input Dimensions

Define the batch size, number of attention heads, sequence length, and head dimension.

These values determine the shape of the 4D input matrices.

```python
# Define input dimensions

batch_size = 2
num_heads = 2
sequence_length = 4
d_k = 4

print("Batch size:", batch_size)
print("Number of heads:", num_heads)
print("Sequence length:", sequence_length)
print("Head dimension:", d_k)
```

## Step 3: Create Q, K and V

Create the Query, Key, and Value matrices.

Each matrix is a 4D tensor with the shape:

(batch_size, num_heads, sequence_length, d_k)

```python
# Create Query matrix

Q = np.random.randn(
    batch_size,
    num_heads,
    sequence_length,
    d_k
)

# Create Key matrix

K = np.random.randn(
    batch_size,
    num_heads,
    sequence_length,
    d_k
)

# Create Value matrix

V = np.random.randn(
    batch_size,
    num_heads,
    sequence_length,
    d_k
)

# Display the shapes

print("Q shape:", Q.shape)
print("K shape:", K.shape)
print("V shape:", V.shape)
```
